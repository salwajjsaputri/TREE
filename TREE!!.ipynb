Membuat Binary Tree Secara Manual

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

# Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
print("\n")
```

Traversal pada BST

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
print("\n")
```

```python
# Tugas 2: Binary Search Tree (BST)
# Data berdasarkan identitas:
# NPM 0051 -> 51
# Tanggal lahir 06 Juni 2006 -> DD=6, MM=6
# ASCII huruf pertama nama depan 'S' = 83
# ASCII huruf kedua nama depan 'a' = 97
# Angka bebas: 10, 25

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return Node(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        elif data > root.data:
            root.right = self.insert(root.right, data)
        return root

# Data unik yang akan dimasukkan ke BST
data_list = [51, 6, 83, 97, 10, 25]  # 6 hanya dimasukkan sekali
bst = BinarySearchTree()
root = None
for value in data_list:
    root = bst.insert(root, value)

# Traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

def preorder(node):
    if node:
        print(node.data, end=' ')
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=' ')

# Pencarian
def search(node, key):
    if node is None or node.data == key:
        return node
    if key < node.data:
        return search(node.left, key)
    return search(node.right, key)

# Output
print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)

# Pencarian
print("\n\n--- Pencarian ---")
key1 = 51  # dua digit terakhir NPM
key2 = 100  # angka tidak dalam list

result1 = search(root, key1)
result2 = search(root, key2)

print(f"{key1} ditemukan dalam tree." if result1 else f"{key1} tidak ditemukan.")
print(f"{key2} ditemukan dalam tree." if result2 else f"{key2} tidak ditemukan.")
```
